### Data Ingestion

```python
# document structure
from langchain_core.documents import Document
```

```python
doc = Document(
    page_content= "main textt content using to create RAG",
    metadata= {
        "source" : "example.txt",
        "pages" : 1 ,
        "author" : "Kabir" , 
        "date_created" : "22-08-2026" 
    }
)
doc
```

```python
#Text Loader
from langchain_community.document_loaders import TextLoader

loaders = TextLoader("../data/python_intro.txt",encoding="utf-8")
document  = loaders.load()
print(document)
```

```python
# Directory loader
from langchain_community.document_loaders import DirectoryLoader

dir_loader = DirectoryLoader(
    "../data",
    glob = "**/*.txt", #pattern to match
    loader_cls = TextLoader , 
    loader_kwargs = { 'encoding' : 'utf-8'},
    show_progress= False
)
documents = dir_loader.load()
documents
```

```python
#pdf loader 
from langchain_community.document_loaders import PyMuPDFLoader , PyPDFLoader

dir_loader = DirectoryLoader(
    "../data",
    glob = "**/*.pdf", #pattern to match
    loader_cls = PyPDFLoader , 
    show_progress= False
)

pdf_documents = dir_loader.load()
pdf_documents
```
